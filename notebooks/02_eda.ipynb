### Exploratory Data Analysis
##### Exploratory Data Analysis (EDA) is an approach used in statistics and data science to understand data before applying formal models.

#### In simple word : we explore the cleaned data to discover patterns, trends and relationships

#### Questions
##### Which Walmart store has the highest total sales?
we use groupby()

```python
# Load pandas
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/Walmart_cleaned.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
# Groupby
# For example we need a single store sales we use groupby()
df.groupby("Store")["Weekly_Sales"].sum()
```

```python
# Find the highest
df.groupby("Store")["Weekly_Sales"].sum().sort_values(ascending=False)
```

```python
# List top 10
top_stores = (
    df.groupby("Store")["Weekly_Sales"].sum().sort_values(ascending=False).head(10)
)
```

```python
top_stores
```

```python
store_sales = (
    df.groupby("Store")["Weekly_Sales"].sum().sort_values(ascending=False)
)
```

```python
store_sales
```

```python
store_sales.head(10)
```

```python
store_sales.tail(10)
```

```python
# Total sales vs average sales
# Does the store with the highest total sales also have the highest average weekly sales?
```

```python
df.groupby("Store")["Weekly_Sales"].mean()
```

```python
store_avg_sales = (
    df.groupby("Store")["Weekly_Sales"].mean().sort_values(ascending=False)
)
```

```python
store_avg_sales.head(10)
```

```python
# How many records does each store have
store_counts =df.groupby("Store")["Weekly_Sales"].count()
```

```python
store_counts.head()
```

```python
store_counts.describe()
```

```python
# Create a summary table
store_summary = (
    df.groupby("Store")["Weekly_Sales"]
    .agg(["sum","mean","min","max","std"])
)
```

```python
store_summary.head()
```

```python
# Sort our summary
store_summary.sort_values(
    "sum",ascending=False
).head(10)
```

```python
# Which 10 walmart stores have the highest total sales?
store_sales = (
    df.groupby("Store")["Weekly_Sales"].sum().sort_values(ascending=False)
)
```

```python
store_sales
```

```python
# Get top 10
top_10_stores = store_sales.head(10)
```

```python
top_10_stores
```

```python
plt.figure(figsize=(10,6))
plt.bar(
    top_10_stores.index.astype(str),
    top_10_stores.values
)
plt.title("Top 10 Walmart Stores by Total Sales")
plt.xlabel("Store")
plt.ylabel("Total Sales")
plt.savefig("../images/eda/top_10_walmart_stores_by_total_sales.png")
plt.show()
```

```python
for store, sales in top_10_stores.items():
    print(store, sales)
```

```python
# Group sales by year
yearly_sales = (
    df.groupby("Year")["Weekly_Sales"]
      .sum()
)
```

```python
yearly_sales
```

```python
# Sort the years
yearly_sales = yearly_sales.sort_index()
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    yearly_sales.index,
    yearly_sales.values,
    marker="o"
)

plt.title("Walmart Total Sales by Year")
plt.xlabel("Year")
plt.ylabel("Total Sales")

plt.xticks(yearly_sales.index)

plt.tight_layout()
plt.savefig("../images/eda/walmart_total_sales_by_year.png")
plt.show()
```

```python
df.Holiday_Flag.value_counts()
```

```python
holiday_summary=(
    df.groupby("Holiday_Flag")["Weekly_Sales"]
    .agg(["count","sum","mean"])
)
holiday_summary.index=[
    "Non-Holiday","Holiday"
]
holiday_summary
```

```python
non_holiday_avg=holiday_summary.loc["Non-Holiday","mean"]
holiday_avg=holiday_summary.loc["Holiday","mean"]
percentage_difference = (
    (holiday_avg - non_holiday_avg)/non_holiday_avg
) * 100
print(f"Holiday sales difference: {percentage_difference:.2f}%")
```

```python
plt.figure(figsize=(8,5))
plt.bar(
    holiday_summary.index,
    holiday_summary["mean"]
)
plt.title("Average weekly sales:Holiday vs Non-Holiday")
plt.xlabel("Week Type")
plt.ylabel("Average Weekly Sales")
plt.savefig("../images/eda/avg_weekly_sales_holiday_vs_non-holiday.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))
bars=plt.bar(
    holiday_summary.index,
    holiday_summary["mean"]
)
plt.title("Average Weekly Sales: Holiday vs Non-Holiday")
plt.xlabel("Week Type")
plt.ylabel("Average Weekly Sales")
for bar in bars:
    value=bar.get_height()
    plt.text(
        bar.get_x()+bar.get_width()/2,
        value,
        f"{value:,.0f}",
        ha="center",
        va="bottom"
    )
plt.savefig("../images/eda/avg_weekly_sales_two-bar_comparison.png")
plt.show()
```
