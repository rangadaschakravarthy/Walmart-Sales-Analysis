#### Coefficient of variation(CV)
##### Formula
$$ CV=Standard Deviation/Average $$

```python
store_performance["CV"] = (
    store_performance["Sales_Std"]
    / store_performance["Average_Weekly_Sales"]
)
```

```python
store_performance.head()
```

```python
volatile_stores = (
    store_performance
    .sort_values("CV", ascending=False)
    .head(10)
)
```

```python
volatile_stores
```

Standard deviation asks:

"How many dollars are the sales fluctuating by?"

CV asks:

"How big is that fluctuation compared with the store's normal sales?"

```python
# Holiday performnace by Store
store_holiday_sales=(
    df.groupby(["Store","Holiday_Flag"])["Weekly_Sales"]
    .mean()
    .reset_index()
)
```

Store 1
   ├── Non-Holiday → average sales
   └── Holiday     → average sales

Store 2
   ├── Non-Holiday → average sales
   └── Holiday     → average sales



```python
store_holiday_sales.head()
```

```python
# How much does this store sell during a holiday week
store_holiday_pivot=store_holiday_sales.pivot(
    index="Store",
    columns="Holiday_Flag",
    values="Weekly_Sales"
)
```

```python
store_holiday_pivot.head()
```

```python
store_holiday_pivot.columns=[
    "Non_Holiday_Avg","Holiday_Avg"
]
```

```python
store_holiday_pivot.head()
```

```python
# Calculate the holiday impact
# How much does average sales change during holidays?
store_holiday_pivot["Holiday_Impact_%"] = (
    (
        store_holiday_pivot["Holiday_Avg"]
        - store_holiday_pivot["Non_Holiday_Avg"]
    )
    / store_holiday_pivot["Non_Holiday_Avg"]
) * 100
```

Suppose Store 20 has:

Non-Holiday average = 2,000,000
Holiday average     = 2,400,000

Then:

Difference = 400,000

Divide by normal sales:

400,000 / 2,000,000 = 0.20

Multiply by 100:

20%

So:

Store 20 has a 20% holiday sales uplift.

```python
# Find stores with the highest holdaiy uplift
holiday_uplift = (
    store_holiday_pivot
    .sort_values("Holiday_Impact_%", ascending=False)
)
```

```python
holiday_uplift.head()
```

```python
# Find stores where holidays don't help
holiday_uplift.tail()
```

```python
# Visualise the top 10
top_holiday_stores = holiday_uplift.head(10)
plt.figure(figsize=(10, 6))
plt.bar(
    top_holiday_stores.index.astype(str),
    top_holiday_stores["Holiday_Impact_%"]
)
plt.title("Top 10 Stores by Holiday Sales Uplift")
plt.xlabel("Store")
plt.ylabel("Holiday Sales Uplift (%)")
plt.savefig("../images/eda/top_10_sales_by_holiday_sales_uplift.png")
plt.show()
```

What is correlation?

Correlation measures how two numerical variables move together.

It ranges from:

-1 ←──────── 0 ────────→ +1
Close to +1

Strong positive relationship.

Example:

X ↑ → Y ↑
Close to −1

Strong negative relationship.

X ↑ → Y ↓
Close to 0

Little/no linear relationship.

X ↑ → Y doesn't consistently change

### Which stores are consistently strong?

Previously, we found:

Store 20 → highest total sales  
Store 33 → lowest total sales  
Store 35 → highest relative volatility

But there's another question:

Does a store with high average sales also have consistent sales?

```python
store_avg = (
    df.groupby("Store")["Weekly_Sales"]
      .mean()
      .sort_values(ascending=False)
)
```

```python
store_avg.head(10)
```

```python
# Create a performance + Stability Table
store_analysis = store_performance.copy()
```

```python
store_analysis["CV"] = (
    store_analysis["Sales_Std"]
    / store_analysis["Average_Weekly_Sales"]
)
```

```python
store_analysis = store_analysis.sort_values(
    "Average_Weekly_Sales",
    ascending=False
)
```

```python
store_analysis[
    [
        "Store",
        "Average_Weekly_Sales",
        "Sales_Std",
        "CV"
    ]
].head()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(
    store_analysis["Average_Weekly_Sales"],
    store_analysis["CV"]
)
plt.xlabel("Average Weekly Sales")
plt.ylabel("Coefficient of Variation (CV)")
plt.title("Store Performance vs Sales Volatility")
plt.savefig("../images/eda/store_performance_vs_sales_volatility.png")
plt.show()
```

### Identify high-performing and stable stores

```python
sales_median = store_analysis["Average_Weekly_Sales"].median()
cv_median = store_analysis["CV"].median()
```

```python
sales_median
```

```python
cv_median
```

```python
high_performing_stable = store_analysis[
    (store_analysis["Average_Weekly_Sales"] > sales_median) &
    (store_analysis["CV"] < cv_median)
]
```

```python
high_performing_stable[
    [
        "Store",
        "Average_Weekly_Sales",
        "Sales_Std",
        "CV"
    ]
].sort_values(
    "Average_Weekly_Sales",
    ascending=False
)
```

```python
plt.figure(figsize=(10, 5))
bars = plt.barh(
    high_performing_stable['Store'].astype(str), 
    high_performing_stable['Average_Weekly_Sales'], 
    xerr=high_performing_stable['Sales_Std'], 
    color='#2b5c8f', 
    ecolor='gray', 
    capsize=4
)

plt.xlabel("Average Weekly Sales")
plt.ylabel("Store Number")
plt.title("Average Weekly Sales by Store (with Standard Deviation)")
plt.savefig('../images/eda/avg_weekly_sales_by_store(with_standard_deviation).png')
plt.show()
```
