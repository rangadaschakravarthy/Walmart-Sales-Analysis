An outlier is a value that is unusually far away from most other values.

#### Detect sales outliers using IQR
##### IQR MEAN: Interquartile Range   
##### It is calculated as:  
$$       IQR=Q3-Q1  $$
##### where:
##### * Q1 =25th percentile
##### * Q3 =75th percentile
##### Then we define:

$$ Lower\ Bound = Q1 - 1.5(IQR) $$
$$ Upper Bound=Q3+1.5(IQR) $$

```python
Q1 = df["Weekly_Sales"].quantile(0.25)
Q3 = df["Weekly_Sales"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 - 1.5 * IQR
print("Q1 : ",Q1)
print("Q3 : ",Q3)
print("IQR : ",IQR)
print("Lower Bound : ",lower_bound)
print("Upper Bound : ",upper_bound)
```

```python
outliers = df[
    (df["Weekly_Sales"] < lower_bound) |
    (df["Weekly_Sales"] > upper_bound)
]
```

```python
outliers.head()
```

```python
outliers.shape
```

Do NOT do this:

df = df.drop(outliers.index)

yet.

Why?

Because we're doing analysis of real Walmart sales.

An unusually high sales value could be completely legitimate.

Instead, we'll investigate the outliers later.

For example:

Store  Date        Weekly Sales
20     2010-12-24  $3,000,000

That might look like an outlier.

But if December 24 is a major holiday period, it may be a perfectly valid observation.


```python
# 10 highest potential sales outliers
outliers.sort_values(
    "Weekly_Sales",
    ascending=False
).head(10)
```

#### Question?

#### Should we remove outliers
No. ❌

This is important.

We found potential outliers.

But we're not going to automatically remove them.

Why?

Because:

Unusual sales
      ≠
Incorrect sales

For example, Walmart might legitimately have very high sales during a holiday.

So we'll keep them.

Later, during EDA, we'll investigate them.

This is actually something you can mention in an interview:

"I used IQR to identify potential outliers, but I did not automatically remove them because extreme sales values could represent genuine business events such as holiday spikes."

That's a good data-analysis answer.
